## <b>Introduction to Sovereign ESG Framework</b>

***

****Environmental, Social, and Governance (ESG)*** framework serves as a critical global benchmark used to evaluate how nations and corporations manage risks and opportunities related to environmental sustainability, societal equity, and institutional integrity. Environmental factors track global carbon footprints and climate resource pressures; Social metrics analyze human capital development and workforce gender dynamics; and Governance examines the institutional structures that drive long-term socioeconomic stability.*

*By leveraging the official World Bank Sovereign ESG Data Portal, the primary aim of this project is to execute an end-to-end exploratory data analysis (EDA) that uncovers how these macro-development indicators intersect across distinct global economies.* 

*To transform this dense institutional framework into actionable insights, this analytical pipeline is engineered using **NumPy** and **Pandas** for robust data manipulation and cleaning, alongside **Matplotlib** and **Seaborn** to build crisp, high-fidelity data visualizations.*

### (V) Handling Missing Values

*The Year columns 2015, 2016, ... 2023 contains missing/ null values for various indicators of different economies.* 

*Each row in the dataset contains values of a particular indicator of a specific Economy spanning over the time period from 2015 to 2023. The missing values are to be replaced with the respective mean value of that particular row over the years 2015 to 2023. This ensures that the dataset remains complete and suitable for further processing.*


```python
# Displaying total number of null values in each column
print("Total number of null values in each column:-")
df_filtered.isnull().sum()
```

```python
# Displaying the tail of dataset to see if there are null values
df_filtered.tail()
```

```python
# Replacing the null values with the mean values of respective rows

year_columns = ["2015","2016","2017","2018","2019","2020","2021","2022","2023"]
row_means = df_filtered[year_columns].mean(axis=1)
df_filtered[year_columns] = df_filtered[year_columns].apply(lambda row: row.fillna(row_means[row.name]), axis=1)
```

```python
# Displaying total number of null values in each column
print("Total number of null values in each column after replacing with mean values:-")
df_filtered.isnull().sum()
```

📝 ***Note:*** *Here there are 318 null values in all columns from 2015 to 2023. This is because there are 318 rows in the dataset where an Economy has absolutely zero data for that entire 2015–2023 window.*

```python
# Displaying the rows with no data points in the entire chosen timeline
completely_empty_rows = df_filtered[df_filtered[year_columns].isnull().all(axis=1)]
completely_empty_rows.head()
```

```python
# Removing the rows with no data points in the 2015-2023 time period
df_filtered = df_filtered.dropna(subset=year_columns, how="all")
```

```python
# Displaying total number of null values in each column
print("Total number of null values in each column after removing the rows with no data points:-")
df_filtered.isnull().sum()
```

```python
# Displaying tail of the dataset to see if the null values still remains in the places earlier
df_filtered.tail()
```

ℹ️ ***Info:-*** *There were exactly 318 rows with no data points in the entire chosen timeline. These rows are removed using ***dropna()***. The total number of null values in each column is displayed again, which shows zero for all columns. That means all the null values are handled properly.*

### (VII) Creating Derived Columns

*In this step, new columns such as **Absolute Change over Time**,  **Percentage Growth over Time**, **Timeline Volatility** and **Average_(2015-23)** are created to transform raw data into much deeper insights*

#### Creating **"Absolute Change over Time"** and **"Percentage Growth over Time"** Columns

```python
# Finding the absolute change in values from 2015 to 2023
df_filtered['Absolute_change_over_time'] = df_filtered['2023'] - df_filtered['2015']

# Creating the "Percentage Growth over Time" column
df_filtered["Percentage_growth_over_time"] = (df_filtered['2023'] - df_filtered['2015'])/(df_filtered['2015'] + 0.00001) * 100
```

📝 ***Note:*** *Here a very small number 0.00001 is added in the denominator to ensure that the denominator does not become zero even if the value in 2015 is zero.*

#### Creating **"Timeline Volatility"** Column

```python
df_filtered['Timeline_volatility'] = df_filtered[year_columns].std(axis=1)
```

#### Creating **'Average_(2015-23)'** Column

```python
df_filtered['Average_(2015-23)'] = df_filtered[year_columns].mean(axis=1)
```

### (b) Fertility rate and Life expectancy in India over time
• ***Visualization Type:-*** ***Combo chart (Bar chart + Line chart)***   
• ***Purpose:-*** *To analyze the Fertility rate(using bar chart) and Life expectancy(using line chart) in India during 2015-2023.   

```python
target_indicators = ["SP.DYN.TFRT.IN","SP.DYN.LE00.IN"]
df_fert_life = df_filtered[(df_filtered["Indicator_code"].isin(target_indicators))&(df_filtered["Economy"]=="India")]   # Filtering data for the selected indicator and Economy
years =  ["2015", "2016", "2017", "2018", "2019", "2020", "2021", "2022", "2023"]
# Reshaping data from wide format to long format
df_long = df_fert_life.melt(
    id_vars = ["Economy","Indicator_code"],
    value_vars = years,
    var_name = "Year",
    value_name = "Value" 
).dropna()
# Pivoting so each indicator gets its own column
df_pivot = df_long.pivot_table(
    index="Year", 
    columns="Indicator_code", 
    values="Value"
).reset_index()

df_pivot = df_pivot.rename(columns={"SP.DYN.TFRT.IN": "Fertility_Rate", "SP.DYN.LE00.IN": "Life_Expectancy"})   # Renaming columns for clarity

# Creating Combo chart
with sns.axes_style("dark"):        # Swiching to dark style from the default darkgrid style
    fig, ax1 = plt.subplots(figsize=(12, 5))
    # Define distinct theme colors
    bar_color = "#2a9d8f"   
    line_color = "#d97706"

    # Bar Chart for Fertility Rate 
    bars = sns.barplot(
        data=df_pivot,
        x="Year",
        y="Fertility_Rate",
        color=bar_color,
        alpha=0.75,
        width=0.45,
        label="Fertility Rate",
        ax=ax1
    )
    ax1.set_xlabel("Year", fontsize=12, weight="bold", labelpad=10,color="black")
    ax1.set_ylabel("Fertility Rate (Births per Woman)", fontsize=11, weight="bold", color=bar_color)
    ax1.tick_params(axis="y", labelcolor=bar_color)
    ax1.set_ylim(0, 3.0)  # Gives perspective to the bar heights

    # Line Chart for Life Expectancy 
    ax2 = ax1.twinx()  # Share the x-axis
    line = ax2.plot(
     df_pivot["Year"],
        df_pivot["Life_Expectancy"],
        color=line_color,
        marker="o",
        linewidth=2.5,
        markersize=7,
        label="Life Expectancy"
    )
    ax2.set_ylabel("Life Expectancy at Birth (Years)", fontsize=11, weight="bold", color=line_color)
    ax2.tick_params(axis="y", labelcolor=line_color)

    # Scale right Y-axis tightly around life expectancy range
    min_life = df_pivot["Life_Expectancy"].min() - 2
    max_life = df_pivot["Life_Expectancy"].max() + 2
    ax2.set_ylim(min_life, max_life)                    # Setting the y limits for line chart based on the minimum value and maximum value

    ax1.legend(loc="upper left")        # Legend for Fertility Rate (Bars) on the left
    ax2.legend(loc="upper right")       # Legend for Life Expectancy (Line) on the right

    plt.title("India: Fertility Rate vs. Life Expectancy Trend (2015-2023)", fontsize=13, weight="bold", style="italic", pad=15)
    # Remove top spine while preserving both left and right y-axes
    sns.despine(ax=ax1, top=True, right=False)
    sns.despine(ax=ax2, top=True, right=False)
    plt.tight_layout()
    plt.show()

```

### (a) Average Mortality rate by Geographic region 
• ***Visualization Type:-*** ***Horizontal Bar chart***   
• ***Purpose:-*** *To analyze the average mortality rate of different geographic regions.*

```python
mortality_rate_indicator = "SH.DYN.MORT"
df_mortality = df_filtered[df_filtered["Indicator_code"] == mortality_rate_indicator]       # Filtering data for the selected indicator
years = ["2015", "2016", "2017", "2018", "2019", "2020", "2021", "2022", "2023"]
# Reshaping data from wide format to long format
df_long = df_mortality.melt(
    id_vars = ["Geographic_region"],
    value_vars = years,
    var_name = "Year",
    value_name = "Mortality_rate"
).dropna(subset=["Mortality_rate","Geographic_region"])
rates = (df_long.groupby("Geographic_region", observed=False)["Mortality_rate"].mean().sort_values(ascending=False))    # Group by region and computing average mortality rate

# Creating Horizontal bar chart
plt.figure(figsize=(12,5))
with sns.axes_style("dark"):        # Swiching to dark style from the default darkgrid style
    ax = sns.barplot(
    x=rates.values,
    y=rates.index,
    palette="RdPu",  
    alpha=0.85
)
    for container in ax.containers:
        ax.bar_label(container, fmt="%.1f", padding=6, fontweight="bold", fontsize=9, color="black")     # Creating labels
    plt.title("Average Mortality Rate by Geographic Region (2015-2023)", fontsize=13, weight="bold",style="italic", pad=20)
    plt.xlabel("Mortality Rate (per 1,000 live births)", fontsize=11, weight="bold", labelpad=10)
    plt.ylabel("Geographic Region", fontsize=11, weight="bold", labelpad=10)
    # Add right-side padding so text labels aren't cut off
    max_val = rates.values.max()
    plt.xlim(0, max_val + 8)
    plt.xticks(fontsize=10)
    plt.yticks(fontsize=10)
    sns.despine(top=True, right=True)
    plt.tight_layout()
    plt.show()
```

So all unnecessary columns were removed. Also the records contained information regarding 195 unique indicatoors. ### <b>💡Recommendations</b>

👉 *CO2 in North America and other regions as well is driven mainly by fossil fuel combustion for electricity, transportation, and industrial manufacturing. This can be reduced by switching to renewable energy (solar, wind and geothermal), expanding to high-density EV infrastructure and Reforestation.*

👉 *Over 75% of North America’s N2O emissions is from agricultural soil management—specifically the overapplication of synthetic nitrogen fertilizers on large-scale commercial crops like corn and wheat. To reduce this, we can follow the following remedies:* 

*(1) Apply nitrogen fertilizer using the 4R Principles: the Right Source, Right Rate, Right Time, and Right Place (2) Utilize chemical nitrification inhibitors mixed into fertilizers (3) Integrate cover crops (like legumes) during off-seasons to fix natural nitrogen in the soil.*

👉 *The long term over usage of Fossil fuel energy is causing Climate Change & Global Warming, Air pollution, damage to public health aand eventually leads to resource depletion. Switching to renewable energy sources, improving public transportation, switching to EV transportaion are few remedies regarding this.*

👉 *High level of water stress occurs when the demand for safe, usable water exceeds the available supply or when poor quality restricts its use. Agriculture is a major component in the fresh water withdrawal. Utilising micro irrigation can save water. Shifting away from water-intensive crops helps to solve the issues faced in arid regions. Advanced water recycling, rain water harvesting, reforestation etc. also helps to solve this issue.*

👉 *Bhutan relies heavily on run-of-the-river hydroelectric power generation. During winter, river levels drop dramatically, cutting hydro output by up to 70–80%. Building floating solar and mountain-based solar/wind farms can help with this issue.This will be help the economy attain 100% access to electricity.*
